### Copilot Activity 2 – Iterative Function

Ask Copilot:

````python
# Activity 2
# Write your iterative Fibonacci function here.
#Function definition with n as parameter to get the nth fibonacci number.


def fibonacci_iteration(n):
    if n <= 1:
        return 0
    else:
        a = 0
        b = 1
        for i in range(2, n):
            t = a
            a = b
            b = t + a
        return b
````

````python
#Function call with 10 as argument to get 10th element of fibonacci series
fibonacci_iteration(10)
````

## 7. Activity 5 – Recursive Fibonacci Function

Now solve the same problem using **recursion**.

A recursive Fibonacci function follows the mathematical definition:

```text
F(0) = 0
F(1) = 1
F(n) = F(n-1) + F(n-2)
```

The function calls itself with smaller values of `n` until it reaches the **base cases**.

### Important concept

A recursive function must have a stopping condition. Without a base case, the function will continue calling itself.

### Copilot Activity 4 – Understand Recursion

Ask Copilot:

> Explain recursion using the Fibonacci function. Clearly identify the base cases and the recursive case. Explain what happens when fibonacci(5) calls smaller Fibonacci functions. Do not write the complete code.

````python
# Activity 5
# Write a recursive Fibonacci function.

def fibonacci_recursion(n):
    if n <= 1:
        return 0
    elif n == 2:
        return 1
    else:
        return fibonacci_recursion(n-1) + fibonacci_recursion(n-2)
````

````python
fibonacci_recursion(10)
````

## 9. Activity 7 – Measure Execution Time

````python
# Activity 7
# Measure and compare execution times.

import time

def measure_execution_time(func, n):
    start_time = time.time()
    result = func(n)
    end_time = time.time()
    return result, end_time - start_time
````

## 14. Optional Challenge – Improve the Recursive Version

The naive recursive function recalculates the same Fibonacci values many times.

Ask Copilot:

> Explain how memoization can improve the performance of recursive Fibonacci. Show the idea conceptually first; do not rewrite my complete practical.

If you understand the explanation, try creating an optional memoized version and compare its execution time with naive recursion.

**This is an optional extension and is not required for the basic practical.**

````python
#Explaination:(Not a program just a general understanding)
#Memoization is a method to elimimate the recalculation of the same function which is already calculated, by just storing the value and lookup fo it whenever needed. This eliminates the biggest drawback of the naive recursion were due to recalculation of same value again and again it slowes down as it grows by (2^n) that is exponential growth.

#General naive recursion growing exponentially:
F(4) = F(3) + F(2)
    F(4) -> needs F(3) first 
        F(3) = F(2) + F(1)
            F(3) -> needs F(2)
                F(2) = F(1) + F(0)
                    F(2) -> now calls F(1) -> base value, returns 1
                    F(2) -> now calls F(0) -> base value, returns 0
                    F(1) -> 1
                    F(0) -> 0
                F(2) retruns 1
            F(3) -> needs F(1) so it calss F(1) -> base value, returns 1
                F(1) -> 1
        F(3) returns 2
    F(4) -> needs F(2) 
        F(2) = F(1) + F(0)
            F(2) -> now calls F(1) -> base value, returns 1
            F(2) -> now calls F(0) -> base value, returns 0
            F(1) -> 1
            F(0) -> 0
        F(2) retruns 1
F(4) -> 3
F(4) returns 3

````

````python
def fib_memo(n, memo={}):
    if n in memo:                  # if n is already calculated lookup for it in stored memory
        return memo[n]
    if n <= 1:                     # base cases: F(0) = 0, F(1) = 1
        return n
    memo[n] = fib_memo(n - 1, memo) + fib_memo(n - 2, memo)   # calculate once and store it
    return memo[n]


n = 20
print("Fibonacci series using memoization:")
for i in range(n):
    print(fib_memo(i), end=" ")
````
