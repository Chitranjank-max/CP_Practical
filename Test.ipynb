```python
#Antoine constants for Benzene (1) and EthylBenzene(2) were temperature in K and Pressure in bar
A1, B1, C1 = 4.72583, 1660.652,	-1.461
A2, B2, C2 = 4.07488, 1419.315,	-60.539

def VLE_calculations(x1, y1):
    x2 = 1 - x1
    y2 = 1 - y1
    k_target = ((y1/x1)/(y2/x2))

    #start with guess temperature by using linear model equation.
    #Benzene B.P = (353.3 ± 0.1)K
    #Ethylbenzene B.P = (409.3 ± 0.4)K
    T = (x1*(353.3)) + (x2*(409.3))

    #To loop the calculations untill ratio of P1_sat/P2_sat gets close enough to k_target
    #with tolerance 10^-6
    tolerance = 1e-6
    while True and 330 < T < 425:
        P1 = 10**(A1 - (B1 / (T + C1)))
        P2 = 10**(A2 - (B2 / (T + C2)))
        current_ratio = (P1/P2)

        #compare the absolute value of current_ratio with k_target
        if abs(current_ratio - k_target) < tolerance:
            break

        diff = current_ratio - k_target
        #print difference comannd
        #adjusting the value of T
        step = diff*2.0
        if diff < k_target:
            T += step
        if diff > k_target:
            T -= step
            

    P = (x1 * P1)/y1
    return T, P

```
