Q2 Batch Reactor.
>Used AI to obtain tabulated data and also for error check analysis.

```python
#Q)A
import math

try:
    C0 = float(input("Enter initial concentration (mol/L): "))
    dt = float(input("Enter time step (min): "))
    K = float(input("Enter rate constant (1/min): "))
    max_conversion = float(input("Enter maximum conversion (%): "))
    max_time = float(input("Enter maximum time (min): "))
    show_table = input("Display the time/concentration/conversion table? (yes/no): ").strip().lower()

    values = (C0, dt, K, max_conversion, max_time)

    if not all(math.isfinite(value) for value in values):
        print("Enter finite numeric values.")
    elif C0 <= 0 or dt <= 0 or K < 0 or max_time < 0:
        print("C0 and time step must be positive; K and maximum time must be non-negative.")
    elif not 0 <= max_conversion <= 100:
        print("Maximum conversion must be between 0 and 100%.")
    elif show_table not in ("yes", "no"):
        print("Please answer yes or no.")
    elif max_time > 0 and K * min(dt, max_time) > 1:
        print("Reduce the time step so K × dt is at most 1.")
    else:
        concentration = C0
        time = 0.0
        conversion = 0.0

        if show_table == "yes":
            print(f"\n{'Time (min)':>12} {'Concentration (mol/L)':>24} {'Conversion (%)':>18}")
            print(f"{time:.3f} {concentration:.3f} {conversion:.3f}")

        while time < max_time and conversion < max_conversion:
            step_size = min(dt, max_time - time)
            concentration += -K * concentration * step_size
            time = min(time + step_size, max_time)
            conversion = ((C0 - concentration) / C0) * 100

            if show_table == "yes":
                print(f"{time:3f} {concentration:6f} {conversion:3f}")

        print(f"\nFinal time: {time:.3f} min")
        print(f"Final concentration: {concentration:.6f} mol/L")
        print(f"Final conversion: {conversion:.3f}%")

        if conversion >= max_conversion:
            print("Maximum conversion reached.")
        else:
            print("Maximum time reached.")

except ValueError:
    print("Please enter valid numeric values.")
```

Q4) Kinetics and gas data.
>I am a second year B.tech Chemical Engineering student generatee a code which has data fram provided from input of table conatining  6 readings of 3 columns that are time(min), Run_1_mol_L, Run_2_mol_L saves it as kinetics.csv without index column and kinetics.json also read CSV back and print its shape and data types.Print the mean and standard deviation and state which run shows more variations.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "time(min)": [0, 10, 20, 30, 40, 50],
    "Run_1_mol_L": [1.20, 0.98, 0.80, 0.66, 0.54, 0.44],
    "Run_2_mol_L": [1.20, 0.80, 0.54, 0.36, 0.24, 0.16]
})

# Save both files
df.to_csv("kinetics.csv", index=False)
df.to_json("kinetics.json", orient="records", indent=2)

# Read the CSV back and inspect it
data = pd.read_csv("kinetics.csv")
print(data)
print("\nShape:", data.shape)
print("\nData types:")
print(data.dtypes)

# Calculate run statistics
runs = ["Run_1_mol_L", "Run_2_mol_L"]
print("\nMean concentrations:")
print(data[runs].mean())
print("\nSample standard deviations:")
print(data[runs].std())

if data["Run_1_mol_L"].std() > data["Run_2_mol_L"].std():
    print("\nRun 1 shows more variation.")
else:
    print("\nRun 2 shows more variation.")

# Plot both runs
plt.figure(figsize=(8, 5))
plt.plot(data["time(min)"], data["Run_1_mol_L"], marker="o", label="Run 1")
plt.plot(data["time(min)"], data["Run_2_mol_L"], marker="s", label="Run 2")
plt.title("Concentration vs Time")
plt.xlabel("Time (min)")
plt.ylabel("Concentration (mol/L)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#Q4d)
import numpy as np
import matplotlib.pyplot as plt

volume = np.array([1, 2, 3, 4, 5], dtype=float)
pressure_bar = np.array([3.02, 1.49, 1.01, 0.76, 0.59])

k = np.mean(pressure_bar * volume)
print(f"Mean value of P × V, k = {k:.3f} bar·L")

reference_volume = np.linspace(volume.min(), volume.max(), 200)
reference_pressure = k / reference_volume

plt.scatter(volume, pressure_bar, color="teal", label="Measured data")
plt.plot(reference_volume, reference_pressure, color="darkorange",
         label=f"Reference curve: P = {k:.3f}/V")

plt.title("Carbon Dioxide Pressure vs Volume")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
