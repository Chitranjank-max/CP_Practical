Q2 Batch Reactor.
>Used AI to obtain tabulated data and also for error check analysis.

```python
#Q)A
import math

try:
    C0 = float(input("Enter initial concentration (mol/L): "))
    dt = float(input("Enter time step (min): "))
    K = float(input("Enter rate constant (1/min): "))
    max_conversion = float(input("Enter maximum conversion (%): "))
    max_time = float(input("Enter maximum time (min): "))
    show_table = input("Display the time/concentration/conversion table? (yes/no): ").strip().lower()

    values = (C0, dt, K, max_conversion, max_time)

    if not all(math.isfinite(value) for value in values):
        print("Enter finite numeric values.")
    elif C0 <= 0 or dt <= 0 or K < 0 or max_time < 0:
        print("C0 and time step must be positive; K and maximum time must be non-negative.")
    elif not 0 <= max_conversion <= 100:
        print("Maximum conversion must be between 0 and 100%.")
    elif show_table not in ("yes", "no"):
        print("Please answer yes or no.")
    elif max_time > 0 and K * min(dt, max_time) > 1:
        print("Reduce the time step so K × dt is at most 1.")
    else:
        concentration = C0
        time = 0.0
        conversion = 0.0

        if show_table == "yes":
            print(f"\n{'Time (min)':>12} {'Concentration (mol/L)':>24} {'Conversion (%)':>18}")
            print(f"{time:.3f} {concentration:.3f} {conversion:.3f}")

        while time < max_time and conversion < max_conversion:
            step_size = min(dt, max_time - time)
            concentration += -K * concentration * step_size
            time = min(time + step_size, max_time)
            conversion = ((C0 - concentration) / C0) * 100

            if show_table == "yes":
                print(f"{time:3f} {concentration:6f} {conversion:3f}")

        print(f"\nFinal time: {time:.3f} min")
        print(f"Final concentration: {concentration:.6f} mol/L")
        print(f"Final conversion: {conversion:.3f}%")

        if conversion >= max_conversion:
            print("Maximum conversion reached.")
        else:
            print("Maximum time reached.")

except ValueError:
    print("Please enter valid numeric values.")
```
