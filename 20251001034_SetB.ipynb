#Q1 Equilibrium conversion checker.

```python
#Equilibrium conversion.
import math

try:
    A = float(input("Provide the value of excess reactant A (mol/L): "))
    B = float(input("Provide the value of limiting reactant B (mol/L): "))
    Kc = float(input("Provide the value of equilibrium constant(Kc):"))

    print("Equilibrium calculator")
    print("----------------------")
    print("Initial Concentrations:")
    print(f"The initial concentration of A(mol/L): {A:.3f}")
    print(f"The initial concentration of B(mol/L): {B:.3f}")
    print(f"The value of equilibrium constant Kc: {Kc:.3f}")

    if Kc > 0 and A >= 0 and B >= 0:
        a = Kc
        b = (-((3*Kc) + 1))
        c = 2*Kc
        conversion = 0
        if ((b**2)-(4*a*c)) >= 0:
            x1 = (-b + ((b**2)-(4*a*c))**0.5)/(2*a)
            x2 = (-b - ((b**2)-(4*a*c))**0.5)/(2*a)
            if 1>= x1 >= 0:
                x = x1
            elif 1>= x2 >= 0:
                x = x2
            elif x1 == x2:
                x = x1
            else:
                print = ("No positive solution for conversion.")
            conversion = (x/B)*100 
            if conversion >= 70:
                print("High conversion rate.")
            elif 40 <= conversion < 70:
                print("Moderate conversion rate.")
            else:
                print("Low conversion rate.")   
            h = ("All calculations were carried out successfully no crash detected.")
            globals().pop("print", None)
            print("-------------------")
            print("Equilibrium output:")
            print(f"The equilibrium concentration of A(mol/L) formed: {A - x:.3f}")
            print(f"The equilibrium concentration of B(mol/L) formed: {B - x:.3f}")
            print(f"The concentration of C(mol/L) formed: {x:.3f}")
            print(f"The equilibrium conversion of B: {conversion:.3f}%")
            print("------------------------------------------------------------")
            print(h)           
        else:
            print = ("Provide valid conversion value.")
            print = ("There was a crash detected")
    else:
        print("Provide positive values of Kc.")
except ValueError:
    print("Provide valid values.")
except NameError:
    print("A variable was not defined")


```
